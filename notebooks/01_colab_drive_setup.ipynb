# 1. Drive bağlantısı ve proje klasörleri

Bu defteri Google Colab'de açın ve **Çalışma zamanı → Tümünü çalıştır** seçeneğini kullanın. Google Drive hesabınıza erişim izni istendiğinde, projenin verilerini saklayacağınız hesabı seçin.

Bu adım yalnızca Drive bağlantısını kurar ve eksik proje klasörlerini oluşturur. Görüntü indirmez. Mevcut dosyaların üzerine yazmaz. GPU gerekmez.

```python
from google.colab import drive

drive.mount('/content/drive')
```

```python
from pathlib import Path

drive_root = Path('/content/drive/MyDrive')
if not drive_root.is_dir():
    raise RuntimeError('Drive bağlantısı kurulamadı. Önce bağlantı hücresini çalıştırın.')

PROJECT_ROOT = drive_root / 'TUBITAK_Urla'
folders = ('catalog', 'raw', 'processed', 'labels', 'datasets', 'models', 'results', 'logs')
for folder in folders:
    (PROJECT_ROOT / folder).mkdir(parents=True, exist_ok=True)

print('Drive bağlantısı ve proje klasörleri hazır.')
print(f'Proje: {PROJECT_ROOT}')
for folder in folders:
    print(f'  {folder}/')
```

## Tamamlandığında

Drive'da **TUBITAK_Urla** klasörünü gördüğünüzde başlangıç adımı tamamdır.

Sonraki adımda çalışma alanını, etiket kaynaklarını ve veri yılını netleştirerek küçük bir görüntü denemesi hazırlayacağız. Etiketlerin yılı kesinleşene kadar toplu indirme başlatmayın.

Colab'in çalışma diski geçicidir; bu defterin oluşturduğu proje klasörleri Google Drive hesabınızda kalır.
